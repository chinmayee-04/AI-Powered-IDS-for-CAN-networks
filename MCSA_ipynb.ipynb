```python
def pie_plot(df, cols_list, rows, cols):
    fig, axes = plt.subplots(rows, cols)
    for ax, col in zip(axes.ravel(), cols_list):
        df[col].value_counts().plot(ax=ax, kind='pie', figsize=(15, 15), fontsize=10, autopct='%1.0f%%')
        ax.set_title(str(col), fontsize = 12)
    plt.show()
```

```python
pie_plot(data_train, ['protocol_type', 'outcome'], 1, 2)
```

```python
def Scaling(df_num, cols):
    std_scaler = RobustScaler()
    std_scaler_temp = std_scaler.fit_transform(df_num)
    std_df = pd.DataFrame(std_scaler_temp, columns =cols)
    return std_df
```

```python
cat_cols = ['is_host_login','protocol_type','service','flag','land', 'logged_in','is_guest_login', 'level', 'outcome']
def preprocess(dataframe):
    df_num = dataframe.drop(cat_cols, axis=1)
    num_cols = df_num.columns
    scaled_df = Scaling(df_num, num_cols)

    dataframe.drop(labels=num_cols, axis="columns", inplace=True)
    dataframe[num_cols] = scaled_df[num_cols]

    dataframe.loc[dataframe['outcome'] == "normal", "outcome"] = 0
    dataframe.loc[dataframe['outcome'] != 0, "outcome"] = 1

    dataframe = pd.get_dummies(dataframe, columns = ['protocol_type', 'service', 'flag'])
    return dataframe
```

```python
scaled_train = preprocess(data_train)
```

```python
x = scaled_train.drop(['outcome', 'level'], axis=1)
x = x.apply(pd.to_numeric, errors='coerce')  # forces float
x = x.values
```

```python
print("dtype:", x.dtype)
print("NaNs:", np.isnan(x).sum() if x.dtype != object else "object dtype")
```

```python
x = scaled_train.drop(['outcome', 'level'] , axis = 1).values
y = scaled_train['outcome'].values
y_reg = scaled_train['level'].values

pca = PCA(n_components=20)
pca = pca.fit(x)
x_reduced = pca.transform(x)
print("Number of original features is {} and of reduced features is {}".format(x.shape[1], x_reduced.shape[1]))

y = y.astype('int')
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
x_train_reduced, x_test_reduced, y_train_reduced, y_test_reduced = train_test_split(x_reduced, y, test_size=0.2, random_state=42)
x_train_reg, x_test_reg, y_train_reg, y_test_reg = train_test_split(x, y_reg, test_size=0.2, random_state=42)
```

```python
kernal_evals = dict()
def evaluate_classification(model, name, X_train, X_test, y_train, y_test):
    train_accuracy = metrics.accuracy_score(y_train, model.predict(X_train))
    test_accuracy = metrics.accuracy_score(y_test, model.predict(X_test))

    train_precision = metrics.precision_score(y_train, model.predict(X_train))
    test_precision = metrics.precision_score(y_test, model.predict(X_test))

    train_recall = metrics.recall_score(y_train, model.predict(X_train))
    test_recall = metrics.recall_score(y_test, model.predict(X_test))

    kernal_evals[str(name)] = [train_accuracy, test_accuracy, train_precision, test_precision, train_recall, test_recall]
    print("Training Accuracy " + str(name) + " {}  Test Accuracy ".format(train_accuracy*100) + str(name) + " {}".format(test_accuracy*100))
    print("Training Precesion " + str(name) + " {}  Test Precesion ".format(train_precision*100) + str(name) + " {}".format(test_precision*100))
    print("Training Recall " + str(name) + " {}  Test Recall ".format(train_recall*100) + str(name) + " {}".format(test_recall*100))

    actual = y_test
    predicted = model.predict(X_test)
    confusion_matrix = metrics.confusion_matrix(actual, predicted)
    cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = ['normal', 'attack'])

    fig, ax = plt.subplots(figsize=(10,10))
    ax.grid(False)
    cm_display.plot(ax=ax)
```

```python
lr = LogisticRegression().fit(x_train, y_train)
evaluate_classification(lr, "Logistic Regression", x_train, x_test, y_train, y_test)
```

```python
knn = KNeighborsClassifier(n_neighbors=20).fit(x_train, y_train)
evaluate_classification(knn, "KNeighborsClassifier", x_train, x_test, y_train, y_test)
```

```python
gnb = GaussianNB().fit(x_train, y_train)
evaluate_classification(gnb, "GaussianNB", x_train, x_test, y_train, y_test)
```

```python
lin_svc = svm.LinearSVC().fit(x_train, y_train)
evaluate_classification(lin_svc, "Linear SVC(LBasedImpl)", x_train, x_test, y_train, y_test)
```

```python
dt = DecisionTreeClassifier(max_depth=3).fit(x_train, y_train)
tdt = DecisionTreeClassifier().fit(x_train, y_train)
evaluate_classification(tdt, "DecisionTreeClassifier", x_train, x_test, y_train, y_test)
```

```python
def f_importances(coef, names, top=-1):
    imp = coef
    imp, names = zip(*sorted(list(zip(imp, names))))

    # Show all features
    if top == -1:
        top = len(names)

    plt.figure(figsize=(10,10))
    plt.barh(range(top), imp[::-1][0:top], align='center')
    plt.yticks(range(top), names[::-1][0:top])
    plt.title('feature importances for Decision Tree')
    plt.show()

features_names = data_train.drop(['outcome', 'level'] , axis = 1)
f_importances(abs(tdt.feature_importances_), features_names, top=18)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=64, activation='relu', input_shape=(x_train.shape[1:]),
                          kernel_regularizer=regularizers.L1L2(l1=1e-5, l2=1e-4),
                          bias_regularizer=regularizers.L2(1e-4),
                          activity_regularizer=regularizers.L2(1e-5)),
    tf.keras.layers.Dropout(0.4),
    tf.keras.layers.Dense(units=128, activation='relu',
                          kernel_regularizer=regularizers.L1L2(l1=1e-5, l2=1e-4),
                          bias_regularizer=regularizers.L2(1e-4),
                          activity_regularizer=regularizers.L2(1e-5)),
    tf.keras.layers.Dropout(0.4),
    tf.keras.layers.Dense(units=512, activation='relu',
                          kernel_regularizer=regularizers.L1L2(l1=1e-5, l2=1e-4),
                          bias_regularizer=regularizers.L2(1e-4),
                          activity_regularizer=regularizers.L2(1e-5)),
    tf.keras.layers.Dropout(0.4),
    tf.keras.layers.Dense(units=128, activation='relu',
                          kernel_regularizer=regularizers.L1L2(l1=1e-5, l2=1e-4),
                          bias_regularizer=regularizers.L2(1e-4),
                          activity_regularizer=regularizers.L2(1e-5)),
    tf.keras.layers.Dropout(0.4),
    tf.keras.layers.Dense(units=1, activation='sigmoid'),
])
```

```python
model.compile(optimizer='adam', loss=tf.keras.losses.BinaryCrossentropy(from_logits=True), metrics=['accuracy'])
```

```python
import joblib

loaded_model = joblib.load('best_random_forest_model.pkl')
print("RandomForestClassifier model loaded successfully.")
```

```python
data_simulation = pd.read_csv("/content/drive/MyDrive/Neurogaurd/KDDTrain+.txt")
data_simulation.columns = columns

data_simulation.loc[data_simulation['outcome'] == "normal", "outcome"] = 'normal'
data_simulation.loc[data_simulation['outcome'] != 'normal', "outcome"] = 'attack'

# Ensure the preprocess function is defined in the current scope or imported if it's in another module
# (Assuming preprocess function from previous cells is available)

# Make a copy to avoid SettingWithCopyWarning if data_simulation is a slice
data_simulation_processed = data_simulation.copy()
scaled_simulation = preprocess(data_simulation_processed)

x_simulation = scaled_simulation.drop(['outcome', 'level'], axis=1)

# Convert all columns to numeric, coercing errors to NaN, then fill NaNs
x_simulation = x_simulation.apply(pd.to_numeric, errors='coerce')
x_simulation = x_simulation.fillna(0) # Or another appropriate strategy, here 0 for simplicity

# Ensure column order matches the original training features
# features_names was defined as data_train.drop(['outcome', 'level'], axis=1) before preprocessing
# We need the column names *after* preprocessing and before dropping outcome/level
# Re-creating the original feature names from scaled_train for proper ordering
original_feature_cols = scaled_train.drop(['outcome', 'level'], axis=1).columns
x_simulation = x_simulation[original_feature_cols]

print("Simulation data prepared successfully.")
print("Shape of x_simulation:", x_simulation.shape)
```

```python
import time

print("Starting real-time traffic simulation...")

# Apply the same PCA transformation used for training the RandomForest model
x_simulation_reduced = pca.transform(x_simulation)

for index, row_reduced in enumerate(x_simulation_reduced):
    # Reshape the single row for prediction
    prediction = loaded_model.predict(row_reduced.reshape(1, -1))

    if prediction[0] == 1:  # Assuming 1 represents 'attack' and 0 represents 'normal'
        original_log = data_simulation.iloc[index]
        print(f"ALERT: Attack detected in log entry {index + 1}! Details: Protocol={original_log['protocol_type']}, Service={original_log['service']}, Flag={original_log['flag']}, Src_Bytes={original_log['src_bytes']}, Dst_Bytes={original_log['dst_bytes']}")

    time.sleep(0.1) # Simulate a traffic rate of 10 logs per second

print("Real-time traffic simulation completed.")
```
